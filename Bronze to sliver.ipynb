```python
dbutils.widgets.text("landing_timestamp","")
dbutils.widgets.text("table_name","")

landing_timestamp= dbutils.widgets.get("landing_timestamp")
table_name= dbutils.widgets.get("table_name")
```

```python
bronzedf= spark.read.table(f"caresyn_syed.bronze.{table_name}").filter(f"landing_timestamp = '{landing_timestamp}'")
```

```python
from pyspark.sql import functions as f

updated_df= bronzedf.withColumn("inserted_timestamp",f.current_timestamp()).withColumn("update_timestamp",f.current_timestamp())
```

```python
updated_df.write.mode('overwrite').saveAsTable(f"caresyn_syed.silver.{table_name}")
```

```python
countdf=updated_df.count()
dbutils.notebook.exit(str(countdf))
```
